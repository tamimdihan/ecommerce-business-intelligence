```python
import pandas as pd
file_path = "../data/raw/Online Retail.xlsx"

df = pd.read_excel(file_path)
```

```python
clean_df = df.copy()
```

```python
print("Rows:", clean_df.shape[0])
print("Columns:", clean_df.shape[1])
print("Exact duplicates:", clean_df.duplicated().sum())
```

```python
clean_df.info()
```

```python
clean_df.head()
```

```python
before = clean_df.shape[0]

clean_df = clean_df.drop_duplicates().copy()

after = clean_df.shape[0]

print("Rows before:", before)
print("Rows after:", after)
print("Rows removed:", before - after)
print("Remaining duplicates:", clean_df.duplicated().sum())
```

```python
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
print(clean_df.columns.tolist())
```

```python
clean_df.columns = [
    "invoice_no",
    "stock_code",
    "description",
    "quantity",
    "invoice_date",
    "unit_price",
    "customer_id",
    "country"
]
```

```python
print(clean_df.columns.tolist())
```

```python
clean_df.dtypes
```

```python
clean_df.info()
```

```python
clean_df[[
    "quantity",
    "invoice_date",
    "unit_price",
    "customer_id"
]].head()
```

```python
clean_df["customer_id"] = clean_df["customer_id"].astype("Int64")
```

```python
clean_df.dtypes
```

```python
print("Missing Customer IDs:", clean_df["customer_id"].isna().sum())
print("Unique Customer IDs:", clean_df["customer_id"].nunique())
```

```python
print("Minimum quantity:", clean_df["quantity"].min())
print("Maximum quantity:", clean_df["quantity"].max())
print("Negative quantity rows:", (clean_df["quantity"] < 0).sum())
print("Zero quantity rows:", (clean_df["quantity"] == 0).sum())
```

```python
print("Minimum unit price:", clean_df["unit_price"].min())
print("Maximum unit price:", clean_df["unit_price"].max())
print("Negative price rows:", (clean_df["unit_price"] < 0).sum())
print("Zero price rows:", (clean_df["unit_price"] == 0).sum())
```

```python
print("Earliest invoice date:", clean_df["invoice_date"].min())
print("Latest invoice date:", clean_df["invoice_date"].max())
print("Missing invoice dates:", clean_df["invoice_date"].isna().sum())
```

```python
clean_df[clean_df["unit_price"] < 0]
```

```python
clean_df["transaction_type"] = "SALE"

clean_df.loc[
    clean_df["unit_price"] < 0,
    "transaction_type"
] = "BAD_DEBT_ADJUSTMENT"

clean_df.loc[
    (clean_df["quantity"] < 0) &
    (~clean_df["invoice_no"].astype(str).str.startswith("C")),
    "transaction_type"
] = "INVENTORY_ADJUSTMENT"

clean_df.loc[
    clean_df["invoice_no"].astype(str).str.startswith("C"),
    "transaction_type"
] = "CANCELLATION"
```

```python
clean_df["transaction_type"].value_counts()
```

```python
pd.crosstab(
    clean_df["transaction_type"],
    clean_df["quantity"] < 0
)
```

```python
clean_df.loc[
    clean_df["unit_price"] == 0,
    ["description", "stock_code", "quantity", "transaction_type"]
].head(30)
```

```python
zero_price = clean_df["unit_price"] == 0

print("Zero-price rows:", zero_price.sum())
print(
    "Zero-price rows with Customer ID:",
    clean_df.loc[zero_price, "customer_id"].notna().sum()
)
print(
    "Zero-price rows without Customer ID:",
    clean_df.loc[zero_price, "customer_id"].isna().sum()
)
```

```python
clean_df["revenue"] = clean_df["quantity"] * clean_df["unit_price"]
```

```python
clean_df[
    ["quantity", "unit_price", "transaction_type", "revenue"]
].head(10)
```

```python
clean_df["sales_revenue"] = clean_df["revenue"].where(
    clean_df["transaction_type"] == "SALE",
    0
)
```

```python
print("Total transaction value:", clean_df["revenue"].sum())
print("Total sales revenue:", clean_df["sales_revenue"].sum())
print(
    "Revenue from cancellations:",
    clean_df.loc[
        clean_df["transaction_type"] == "CANCELLATION",
        "revenue"
    ].sum()
)
print(
    "Revenue from inventory adjustments:",
    clean_df.loc[
        clean_df["transaction_type"] == "INVENTORY_ADJUSTMENT",
        "revenue"
    ].sum()
)
print(
    "Revenue from bad-debt adjustments:",
    clean_df.loc[
        clean_df["transaction_type"] == "BAD_DEBT_ADJUSTMENT",
        "revenue"
    ].sum()
)
```
